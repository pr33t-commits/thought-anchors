```python
result_df = pd.concat([
    pd.DataFrame(array).assign(problem_id=problem_id, chunk_id=chunk_id).reset_index().rename(columns = {'index':'token_id'})
    for problem_id, chunks in data.items()
    for chunk_id, array in chunks.items()
], ignore_index=True)
```

```python
result_df.shape
```

```python
distance_threshold = 0.5
depth_threshold = 0.9
num_layers = 32
```

```python
# Step 1: Pivot the dataframe - convert layer columns to rows
# Identify layer columns (numeric columns 0-31)
layer_cols = [str(i) for i in range(num_layers)]

result_df_melted = result_df.melt(
    id_vars=['problem_id', 'chunk_id', 'token_id'],
    value_vars=layer_cols,
    var_name='layer_num',
    value_name='distance'
)

# Create layer names: layer_1, layer_2, ..., layer_32
result_df_melted['layer'] = 'layer_' + (result_df_melted['layer_num'].astype(int) + 1).astype(str)
result_df_melted = result_df_melted.drop('layer_num', axis=1)

result_df_melted.head()
```

```python
# Step 2: For each (problem, chunk, token), sort by layer and compute cumulative minimum
result_df_melted['layer_idx'] = result_df_melted['layer'].str.replace('layer_', '').astype(int)
result_df_melted = result_df_melted.sort_values(['problem_id', 'chunk_id', 'token_id', 'layer_idx'])

# Compute cumulative minimum distance for each token
result_df_melted['distance'] = result_df_melted.groupby(['problem_id', 'chunk_id', 'token_id'])['distance'].cummin()

result_df_melted.head(40)
```

```python
# Step 3: For each (problem, chunk, token), find first layer where distance < threshold
# Then check if that layer index >= depth_threshold * num_layers

def get_is_dtt(group):
    """
    For a given token, find the first layer where distance < distance_threshold.
    If that layer index >= depth_threshold * num_layers, return 1, else 0.
    If no layer crosses the threshold, return 0.
    """
    below_threshold = group[group['distance'] < distance_threshold]
    if len(below_threshold) == 0:
        # No layer crosses threshold, so is_dtt = 0
        return 0
    else:
        # Get first layer that crosses threshold
        first_layer_idx = below_threshold.iloc[0]['layer_idx']
        threshold_layer = depth_threshold * num_layers
        return 1 if first_layer_idx >= threshold_layer else 0

# Group by problem, chunk, token and apply the function
is_dtt_df = result_df_melted.groupby(['problem_id', 'chunk_id', 'token_id']).apply(get_is_dtt).reset_index()
is_dtt_df.columns = ['problem_id', 'chunk_id', 'token_id', 'is_dtt']

is_dtt_df.head(20)
```

```python
# Step 4: Load chunks_labeled and extract importance measures
# Assuming chunks_labeled is loaded from a JSON file
# with open("path/to/chunks_labeled.json") as f:
#     chunks_labeled = json.load(f)

# Extract the four importance measures and create a dataframe
importance_data = []
for chunk in chunks_labeled:
    importance_data.append({
        'chunk_id': chunk['chunk_idx'],
        'resampling_importance_accuracy': chunk['resampling_importance_accuracy'],
        'resampling_importance_kl': chunk['resampling_importance_kl'],
        'counterfactual_importance_accuracy': chunk['counterfactual_importance_accuracy'],
        'counterfactual_importance_kl': chunk['counterfactual_importance_kl']
    })

importance_df = pd.DataFrame(importance_data)
importance_df.head()
```

```python
# Step 5: Merge importance measures with is_dtt_df on problem_id and chunk_id
# Add problem_id to importance_df (assuming problem_id can be extracted from the data source or passed in)
# For now, assuming you'll set problem_id appropriately
# importance_df['problem_id'] = problem_id  # Set this based on your data source

# Merge on problem_id and chunk_id
result_merged = is_dtt_df.merge(
    importance_df,
    on=['chunk_id'],
    how='left'
)

result_merged.head()
```
